```python

def palindrome_number(n):
    original=n
    r_n=0
    while n>0:
        last_digit=n%10
        r_n=10*r_n+last_digit
        n=n//10
    if r_n==original:
        print("palindrome")
    else:
        print("not a palindrome")
palindrome_number(1221)  

#or

def reverse_number(n):
    r_n = 0

    while n > 0:
        digit = n % 10
        r_n = r_n * 10 + digit
        n = n // 10

    return r_n


def is_palindrome(n):
    original = n
    r_n = reverse_number(n)

    if original == r_n:
        return True
    else:
        return False


n = int(input())

if is_palindrome(n):
    print("Palindrome")
else:
    print("Not Palindrome")

   """ Time: O(log n)
Space: O(1)"""
        
```
